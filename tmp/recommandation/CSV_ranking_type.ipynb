```python
import pandas as pd
import numpy as np
```

```python
user_profiles = pd.read_csv('user_profiles.csv')
movies = pd.read_csv('movies.csv')
```

```python
genre_cols = movies.columns.difference(
    ["movie_id", "title", "rates", "nb_rates", "mean", "year", "genre_list"]
)
mean_cols = [f"mean_{genre}" for genre in genre_cols]

type_profiles = (
    user_profiles
    .groupby("id_type")[mean_cols]
    .mean()
    .reset_index()
)

movie_matrix = movies[genre_cols].values
type_matrix = type_profiles[mean_cols].values

scores = np.dot(type_matrix, movie_matrix.T)
```

```python
type_recommendations = (
    pd.DataFrame(
        scores,
        index=type_profiles["id_type"],
        columns=movies["movie_id"]
    )
    .stack()
    .reset_index()
)

type_recommendations.columns = ["id_type", "movie_id", "score"]

type_recommendations = type_recommendations.sort_values(
    ["id_type", "score"],
    ascending=[True, False]
)
type_recommendations.head()
```

```python
type_recommendations.to_csv("type_recommendations.csv", index=False)
```
